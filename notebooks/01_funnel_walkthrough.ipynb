# Worked example: the purchase funnel

This notebook repeats `sql/01_worked_example_funnel.sql` in pandas, so you can see the same four numbers in a second tool.

It prints the same four steps as `outputs/worked_example_funnel.csv`.

Run this notebook from the repository root (or from `notebooks/`; the cells handle both).


```python
from pathlib import Path
import runpy
import pandas as pd

root = Path.cwd()
if not (root / "data").exists():
    root = root.parent
script = root / "notebooks" / "01_funnel_walkthrough.py"
runpy.run_path(str(script), run_name="__main__")
sessions = pd.read_csv(root / "data" / "processed" / "funnel_sessions.csv")
sessions[["session_date", "device", "channel", "landing_page", "flag_view_item", "flag_closed_add_to_cart", "flag_closed_purchase"]].head(8)
```

## What the columns mean

- One row is one session, not one user and not one event.
- `flag_view_item` is 1 if that session viewed a product.
- `flag_closed_add_to_cart` is 1 only if a product view happened at or before the add to cart.
- `flag_closed_purchase` is 1 only if that cart happened at or before the purchase.
- `flag_purchase` is the open version: any purchase, in any order. It is larger. The difference is sessions that purchased without a view, then a cart, then the purchase, in that order.
